### 3 Way Conversation

```python
# Prompts

message_bella = "Bella: Hi Mom! Hi Alice! It's nice to see you after so long. I should be able to stay here for the next week."
message_betsy = "Betsy: Bella! We missed you so much. Welcome home! I'm glad you're doing so well for yourself in the city."
message_alice = "Alice: Hi Bella, welcome home! Can you tell me more about your life? I'm struggling to figure out what I want to do in the future right now."

conversation = f"{message_bella}\n{message_betsy}\n{message_alice}"

system_prompt_alice = """
You are Alice, a young teenage girl who's trying to find her way in life. You are currently feeling very lost and unsure of
what you want to do with your life and feel very uncertain about the future. You are in conversation with your Mother, Betsy, and\
your yound adult older sister, Bella. 
"""

system_prompt_betsy = """
You are Betsy, a mother of two daughters far apart in age, Alice and Bella. You want your children to find happiness and
fulfillment in your life, but you also want them to work hard enough that they are successful and can support themselves
and be independent when they are older. You are in conversation with your younger daughter, Alice, and your older daughter,
Bella.
"""

system_prompt_bella = """
You are Bella, a young adult living with her boyfriend in a big city, fully financially independent from her family. You
feel that you have worked very hard and gone through many challenges to get to where you are. You are in converstion with
your younger sister, Alice, and your mother, Bella.
"""

def user_prompt_bella(conversation):

    bella = f"""
    You are Bella, in conversation with Alice and Betsy.
    The conversation so far is as follows: {conversation}
    Now with this, respond with what you would like to say next, as Bella.
    """
    return bella

def user_prompt_betsy(conversation):

    betsy = f"""
    You are Betsy, in conversation with Alice and Bella.
    The conversation so far is as follows: {conversation}
    Now with this, respond with what you would like to say next, as Betsy.
    """

    return betsy

def user_prompt_alice(conversation):

    alice = f"""
    You are Alice, in conversation with Betsy and Bella.
    The conversation so far is as follows: {conversation}
    Now with this, respond with what you would like to say next, as Alice.
    """

    return alice
```

```python
def call_bella(conversation):

    response = claude1.messages.create(
        
    model=MODEL_CLAUDE,
    max_tokens=200,
    system=system_prompt_bella,
    messages=[{"role": "user", "content": user_prompt_bella(conversation)}])

    reply = f"\n\nBella: {response.content[0].text}"
    display(Markdown(reply))

    conversation += reply

    return conversation
```

```python
def call_betsy(conversation):

    response = llama.chat.completions.create(
        
    model=MODEL_LLAMA,
    messages=[{"role": "system", "content": system_prompt_betsy},
              {"role": "user", "content": user_prompt_betsy(conversation)}]
              )

    reply = f"\n\nBetsy: {response.choices[0].message.content}"
    display(Markdown(reply))

    conversation += reply
    return conversation
```

```python
def call_alice(conversation):

    response = claude2.messages.create(
        
    model=MODEL_CLAUDE,
    max_tokens=200,
    system=system_prompt_alice,
    messages=[{"role": "user", "content": user_prompt_alice(conversation)}])

    reply = f"\n\nAlice: {response.content[0].text}"
    display(Markdown(reply))

    conversation += reply

    return conversation
```

```python
def full_convo(conversation):
    print(conversation)
    for i in range(5):
        conversation = call_bella(conversation)
        conversation = call_betsy(conversation)
        conversation = call_alice(conversation)
```
