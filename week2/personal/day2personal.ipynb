```python
import os
from dotenv import load_dotenv
from openai import OpenAI
import anthropic
import gradio as gr
```

```python
# Models

load_dotenv(override=True)

MODEL_CLAUDE = 'claude-haiku-4-5'
MODEL_LLAMA = 'llama3.2'

# Claude

claude = anthropic.Anthropic()

#  Llama

llama = OpenAI(base_url="http://localhost:11434/v1", api_key="ollama")
```

```python
def message_llama(prompt):
    response = llama.chat.completions.create(
        model = MODEL_LLAMA,
        messages = [{"role": "system", "content": "You are a helpful assistant that responds in Markdown without code blocks."},
                    {"role": "user", "content": prompt}
        ])

    reply = response.choices[0].message.content

    return reply
```

```python
gr.Interface(message_llama, inputs="textbox", outputs="textbox", flagging_mode="never").launch(inbrowser = True)
```
